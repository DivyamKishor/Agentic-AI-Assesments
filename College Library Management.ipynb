```python
print("===================================")
print("   Welcome to College Library Bot")
print("===================================")

print("Type 'exit' anytime to leave.\n")

# Simple library database
books = {
    "python": True,
    "data structures": True,
    "operating systems": False,
    "computer networks": True,
    "database management": True
}

while True:
    user = input("You: ").lower()

    if user == "exit":
        print("Bot: Thank you for using the Library Bot. Goodbye!")
        break

    elif "hello" in user or "hi" in user:
        print("Bot: Hello! How can I help you?")

    elif "timing" in user or "open" in user:
        print("Bot: The library is open from 9:00 AM to 6:00 PM, Monday to Saturday.")

    elif "books" in user or "available" in user:
        print("Bot: Here are the books in our library:")
        for book in books:
            status = "Available" if books[book] else "Not Available"
            print("-", book.title(), ":", status)

    elif "borrow" in user:
        book = input("Bot: Which book do you want to borrow? ").lower()

        if book in books:
            if books[book]:
                books[book] = False
                print("Bot: You have successfully borrowed", book.title())
            else:
                print("Bot: Sorry, this book is currently unavailable.")
        else:
            print("Bot: Sorry, this book is not in our library.")

    elif "return" in user:
        book = input("Bot: Which book do you want to return? ").lower()

        if book in books:
            books[book] = True
            print("Bot: Thank you! You have successfully returned", book.title())
        else:
            print("Bot: This book does not belong to our library.")

    elif "help" in user:
        print("Bot: You can ask me about:")
        print("- Library timings")
        print("- Available books")
        print("- Borrowing a book")
        print("- Returning a book")

    else:
        print("Bot: Sorry, I don't understand that.")
        print("Bot: Type 'help' to see what I can do.")
```
